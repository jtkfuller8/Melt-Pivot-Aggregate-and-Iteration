```python
%pip install ucimlrepo pandas



from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
breast_cancer = fetch_ucirepo(id=14) 
  
# data (as pandas dataframes) 
X = breast_cancer.data.features 
y = breast_cancer.data.targets 
  
# metadata 
print(breast_cancer.metadata) 
  
# variable information 
print(breast_cancer.variables) 
```

```python
print(X)
X.describe(include='all')
```

```python
import pandas as pd
```

```python

#MELT

df_melted = pd.melt(
    X,
    id_vars=['deg-malig', 'tumor-size', 'inv-nodes',
             'node-caps', 'breast', 'breast-quad','irradiat'],
    value_vars=['age', 'menopause'],
    var_name='variable',
    value_name='value'
)

print(df_melted)

# Melt reorganizes the age and menopause columns into two columns.
# The resulting dataset can be used for further analysis of these two variables (age and menopause) when it comes to the severity of tumor/breast cancer.
```

```python

#PIVOT
X = X.copy()
X['row_id'] = range(len(X))

pivot_df = X.pivot(
    index=['row_id','deg-malig', 'tumor-size', 'inv-nodes',
           'node-caps', 'menopause','age'],
    columns='breast',
    values='breast-quad'
)
## Used `pivot()` with a unique row ID to prevent duplicate index errors and display the specific breast quadrant where each tumor is located.
print(pivot_df)
```

```python
#AGGREGATE
avg_malig = X['deg-malig'].mean()
print(avg_malig)
#This will tell the average degree of malignancy in this overall dataset

malig_stats = X['deg-malig'].agg(['mean', 'median', 'min', 'max', 'count'])
# if we want all the stats that could be associated with this column, we try this
print(malig_stats)
```

```python
#ITERATION
unique_sections = X['menopause'].unique()
print(unique_sections)

for idx in X.index:
    current_value = X.loc[idx, 'menopause']

    if current_value == 'premeno':
        X.loc[idx, 'menopause'] = 'premenopausal'

    elif current_value == 'lt40':
        X.loc[idx, 'menopause'] = 'menopause under 40'

    elif current_value == 'ge40':
        X.loc[idx, 'menopause'] = 'menopause 40 or older'

#This serves to iterate through the menopause column and change the names to something more human readable and understandable
print(X)

```

```python
#GROUP BY
X.groupby('menopause')['deg-malig'].mean()
#This groups the column menopause and calculate the average degree of malignancy per menopause subject.
```
